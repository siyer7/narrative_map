```python
def get_extrapolatedPR(event_stims):
    
    x_pred = event_stims.copy()
    for i in range(2, event_stims.shape[1]):
        # extrapolate the next value based on the previous two values
        x_pred[:,i] = (x_pred[:,i-1]) + (x_pred[:,i-1]-x_pred[:,i-2])
    return x_pred

def get_intrapolatedPR(event_stims):
    
    x_pred = event_stims.copy()
    for i in range(1, event_stims.shape[1]-1):
        # intrapolate the current value based on the previous value and the difference between the last and first values
        x_pred[:,i] = (x_pred[:,i-1]) + (x_pred[:,-1]-x_pred[:,0])
    return x_pred

def get_pred_recon_err(stim, bounds, metric='01'):

    # create & populate event_idx -> event_stims
    event2stims = {i: [] for i in range(len(bounds))}  # dict of len=len(bounds)
    for bound_idx in range(len(bounds)):
        event2stims[bound_idx] = stim[:,bounds[bound_idx]:bounds[bound_idx+1]] if bound_idx+1 < len(bounds) else stim[:,bounds[bound_idx]:]
    print(f'events (n,t): {[np.shape(val) for key, val in event2stims.items()]}')

    # per event, compute PRE; then aggregate
    pred_recon_err = []
    for event_idx, event_stims in event2stims.items():

        if metric == '01':  x_true = event_stims[:,2:]; x_pred = get_extrapolatedPR(event_stims)[:,2:]
        elif metric == '0-1': x_true = event_stims[:,1:-1]; x_pred = get_intrapolatedPR(event_stims)[:,1:-1]

        pred_recon_err.extend(np.linalg.norm(x_pred-x_true, axis=0))
    return np.linalg.norm(pred_recon_err)

seed=0; event_len_var=0
metric='0-1'
stim_info = generate_stim(stim_dim=50, event_dim=5, n_stims=100, min_events=5, mean_events=5, max_events=5, event_len_var=event_len_var, stim_noise=0.05, prop_related_events=0.2, related_event_template_noise=0.05, seed=seed)
pred_recon_err = get_pred_recon_err(stim_info['stim'], stim_info['event_boundaries'], metric=metric)
print(pred_recon_err)
```
